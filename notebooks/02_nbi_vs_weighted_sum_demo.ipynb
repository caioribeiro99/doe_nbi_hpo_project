# NBI vs weighted-sum scalarization (demo)

Reproduces the figure that motivates the methodological correction (METHODOLOGY_DECISIONS D1).

Two 2D problems with curved Pareto fronts: NBI gives uniform spacing, weighted sum clusters near the extremes.

```python
import sys, pathlib
REPO = pathlib.Path('.').resolve()
if (REPO/'src').exists():
    sys.path.insert(0, str(REPO/'src'))
import numpy as np
from doe_xgb.nbi_core import NBIConfig, run_nbi
from doe_xgb.simplex import generate_simplex_lattice
from doe_xgb.scalarization import run_nbi_weighted_sum, predict_from_coeffs
```

```python
def f1(x): return float((x[0]-0.0)**2 + 4.0*x[1]**2)
def f2(x): return float((x[0]-1.0)**2 + 4.0*x[1]**2)
bounds = np.array([[-0.5, 1.5], [-0.5, 0.5]])
weights = np.column_stack([np.linspace(0,1,11), 1-np.linspace(0,1,11)])[:, ::-1]
cfg = NBIConfig(objective_count=2, bounds=bounds, n_starts=4, seed=0)
run = run_nbi([f1, f2], weights, cfg)
F_nbi = np.array([c.F_at_x for c in run.candidates])
F_nbi
```
